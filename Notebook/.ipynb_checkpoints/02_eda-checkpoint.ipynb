```python
import pandas as pd
from sqlalchemy import create_engine
from utils.database import get_engine
import seaborn as sns
import matplotlib.pyplot as plt



engine = get_engine()
```

```python
df = pd.read_sql(
    "SELECT * FROM vw_profit_analysis",
    engine
)

df.head()
df.columns
```

```python
df.describe()
```

```python
# Total Business KPIs
total_revenue = df["revenue_generated"].sum()
total_profit = df["profit"].sum()

print("Total Revenue:", total_revenue)
print("Total Profit:", total_profit)
```

```python
# Profit by Product Type
df.groupby("product_type")["profit"].sum().sort_values(ascending=False)
```

```python
#Top 10 Profit Product
df.groupby("sku")["profit"].sum().sort_values(ascending=False).head(10)
```

```python
#Low Profit Product
df[df["profit"] < df["profit"].quantile(0.2)][["sku", "profit"]].sort_values("profit")
```

```python
# Revenue vs Profit Relationship
sns.scatterplot(data=df, x="revenue_generated", y="profit")
plt.title("Revenue vs Profit")
plt.show()
```

```python
# Manufacturing Cost vs Profit
sns.scatterplot(data=df, x="manufacturing_costs", y="profit")
plt.title("Manufacturing Cost vs Profit")
plt.show()
```

```python
# Shipping Cost vs Profit
sns.scatterplot(data=df, x="shipping_costs", y="profit")
plt.title("Shipping Cost vs Profit")
plt.show()
```

## Key Insights – Profit Analysis

### 1. Overall Performance
- Total Revenue: 577,604
- Total Profit: 519,398
- Profit margin is approximately ~90%, indicating high profitability in the dataset.

### 2. Product Performance
- Skincare is the most profitable product category.
- Haircare and Cosmetics contribute moderately but are lower than Skincare.

### 3. SKU-Level Insights
- A small number of SKUs (e.g: SKU2, SKU38, SKU67) contribute high profit.
- Several SKUs generate significantly lower profit, indicating imbalance in product contribution.

### 4. Revenue vs Profit Relationship
- Strong positive correlation observed between revenue and profit.
- Higher revenue products generally lead to higher profit.

### 5. Cost Impact Analysis
- Manufacturing cost and shipping cost show weak correlation with profit.
- Profit is primarily driven by revenue rather than cost variation.

### 6. Business Interpretation
- The dataset shows characteristics of a Pareto distribution (80/20 rule).
- A small subset of products drives majority of business profit.
