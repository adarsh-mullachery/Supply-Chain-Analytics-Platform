```python
import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

#Load csv file
df = pd.read_csv("DataSource/supply_chain_data.csv")

#Converting Col Name
df.columns = (
    df.columns
      .str.strip()           
      .str.lower()     
      .str.replace(" ", "_") 
)

# Connect to MySQL server
connection_url = URL.create(
    drivername="mysql+pymysql",
    username="root",
    password="password@123",
    host="localhost",
    database="SupplyChain"
)

engine = create_engine(connection_url)

#Load to MySQL
df.to_sql(
    name="supply_chain",
    con=engine,
    if_exists="replace",
    index=False
)
```
